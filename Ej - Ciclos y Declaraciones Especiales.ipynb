# Ej Modulo 11 - Ciclos y Declaraciones Especiales

```python
import time

def countup(time_sec):
    current = 0
    while current <= time_sec:
        mins, secs = divmod(current, 60)
        timeformat = '{:02d}:{:02d}'.format(mins, secs)
        print(timeformat, end='\r')
        time.sleep(1)
        current += 1

    print("¡Tiempo completado!")

# Contar hasta 90 segundos
countup(90)
```

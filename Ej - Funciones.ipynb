```python
def checkout_cart(cart):
    def calculate_total(items):
        total = sum(price for _, price in items)
        iva = total * 0.16
        return total + iva

    total_price = calculate_total(cart)
    return f"Total a pagar (IVA incluido): ${total_price:.2f}"

# Ejemplo de uso
cart_1 = [("camiseta", 200), ("tenis", 850), ("pantalón", 500)]
cart_2 = [("libro", 300), ("mochila", 450)]

print(checkout_cart(cart_1))
print(checkout_cart(cart_2))
```

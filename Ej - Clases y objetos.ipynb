# Ej Modulo 13 - Clases y objetos

```python
class Usuario:
    def __init__(self, nombre, correo, rol):
        self.nombre = nombre
        self.correo = correo
        self.rol = rol

    def saludar(self):
        return f"Hola, soy {self.nombre}"

    def obtener_correo(self):
        return self.correo

    def obtener_rol(self):
        return self.rol


class Cliente(Usuario):
    def __init__(self, nombre, correo, rol):
        super().__init__(nombre, correo, rol)
        self.carrito = []

    def agregar_al_carrito(self, producto):
        self.carrito.append(producto)

    def ver_carrito(self):
        return self.carrito


usuario1 = Usuario("María", "maria@email.com", "admin")
print(usuario1.saludar())
print(usuario1.obtener_correo())
print(usuario1.obtener_rol())

cliente1 = Cliente("Luis", "luis@email.com", "cliente")
cliente1.agregar_al_carrito("camisa")
cliente1.agregar_al_carrito("tenis")

print(cliente1.saludar())
print(cliente1.ver_carrito())
```
